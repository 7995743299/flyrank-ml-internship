## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

```python
Weak Picks and Leakage Check

Weak Picks

This baseline uses only three simple rules (content age, CTR, and impressions). As a result, some recommendations may not actually require a refresh.

Examples include:
- Pages with low CTR because of search intent rather than poor content.
- Pages ranking for highly competitive keywords.
- Seasonal pages that naturally have lower engagement during certain periods.
- Pages with zero CTR because of inaccurate or limited tracking.

 Potential Data Leakage

No future information was used when computing the baseline score. The scoring relies only on existing historical features available at the time of evaluation:

- Content age
- Historical impressions
- Historical CTR

No post-refresh performance metrics or future outcomes were included, so the baseline avoids target leakage.
```
